```python
#multithreading
#sub threads running the tasks
import threading,time
class Multiplication:
  @classmethod
  def getTable(cls,n):
    print("Printing {} table:".format(n))
    print("-"*70)
    Multiplication().table(n)
  def table(self,n):
    print("current thread name",threading.current_thread().name)
    for i in range(1,11):
      print("{} x {} = {}".format(n,i,n*i))
      time.sleep(1)
    print("-"*70)
start = time.time()
t1 = threading.Thread(target=Multiplication.getTable,args=(12,))
t2 = threading.Thread(target=Multiplication.getTable,args=(25,))
t1.start()
t2.start()
t1.join()
t2.join()
end = time.time()
print("time taken",end-start)
```

```python
class MultiThreading:
  def __init__(self,n):
    self.n = n
  def table(self):
    for i in range(1,11):
      print("{} x {} = {}".format(self.n,i,self.n*i))
      time.sleep(1)
    print("-"*70)
start = time.time()
t1 = threading.Thread(target=MultiThreading(12).table)
t2 = threading.Thread(target=MultiThreading(25).table)
t1.start()
t2.start()
t1.join()
t2.join()
end = time.time()
print("time taken",end-start)
```
