```python
import numpy as np
```

```python
arr = np.array([-3.14,3.121,12.99,25.2512])
```

```python
#around
a = np.around(arr)
for ele in a:
  print(ele,end=' ')
print()
b = np.around(arr,2)
print(b)
```

```python
#floor
#arr = [-3.14,3.121,12.99,25.2512]
a = np.floor(arr)
for ele in a:
  print(ele,end=' ')
print()
#ceil
a = np.ceil(arr)
for ele in a:
  print(ele,end=' ')
print()
```

```python
#gcd
n1 = 12
n2 = 22
gcd = np.gcd(n1,n2)
print(gcd)
#gcd of nd array object
arr = np.array([12,8,14,22])
gcd = np.gcd.reduce(arr)
print(gcd)
```

```python
#lcm
n1 = 12
n2 = 25
lcm = np.lcm(n1,n2)
print(lcm)
#lcm of nd array object
arr = np.array([12,8,14,22])
lcm = np.lcm.reduce(arr)
print(lcm)
```

```python
#descrete differences
arr = np.array([10,20,30,40,50])
dd = np.diff(arr)
print(dd)
#finding descrete difference n times
arr = np.array([10,100,1000,10000,100000,1000000])
dd = np.diff(arr,4)
print(dd)
#decrete differences on 2d array
arr = np.array([12,25,8,11,18,17,21,22,7]).reshape(3,3)
dd = np.diff(arr,axis=0,n=2)
print(dd)
```
