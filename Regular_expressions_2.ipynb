# Regular Expression (Regex) Cheat Sheet

## 1. Character Matching

| Category | Regex | Meaning | Example |
|:---|:---:|:---|:---|
| Characters | `.` | Any single character | `c.t` → `cat`, `cut` |
| Characters | `[abc]` | `a`, `b`, or `c` | `[abc]` → `a` |
| Characters | `[a-z]` | Any lowercase letter | `[a-z]` → `a` |
| Characters | `[A-Z]` | Any uppercase letter | `[A-Z]` → `A` |
| Characters | `[0-9]` | Any digit | `[0-9]` → `5` |
| Characters | `[^abc]` | Anything except `a`, `b`, `c` | `[^abc]` → `d` |
| Characters | `[^0-9]` | Anything except a digit | `[^0-9]` → `a` |

---

## 2. Character Shortcuts

| Category | Regex | Meaning | Example |
|:---|:---:|:---|:---|
| Shortcut | `\d` | Any digit | `123` |
| Shortcut | `\D` | Any non-digit | `abc` |
| Shortcut | `\s` | Any whitespace | Space, tab, newline |
| Shortcut | `\S` | Any non-whitespace | `A`, `5`, `@` |
| Shortcut | `\w` | Word character | `A`, `5`, `_` |
| Shortcut | `\W` | Non-word character | `@`, `#`, `!` |

---

## 3. Quantifiers

| Category | Regex | Meaning | Example |
|:---|:---:|:---|:---|
| Quantifier | `*` | 0 or more | `a*` |
| Quantifier | `+` | 1 or more | `a+` |
| Quantifier | `?` | 0 or 1 | `a?` |
| Quantifier | `{n}` | Exactly n times | `\d{3}` |
| Quantifier | `{n,}` | n or more times | `\d{3,}` |
| Quantifier | `{n,m}` | Between n and m times | `\d{3,5}` |

---

## 4. Position / Anchors

| Category | Regex | Meaning | Example |
|:---|:---:|:---|:---|
| Position | `^` | Beginning of string/line | `^Hello` |
| Position | `$` | End of string/line | `world$` |
| Position | `\b` | Word boundary | `\bcat\b` |
| Position | `\B` | Not a word boundary | `\Bcat` |

---

## 5. Alternation and Groups

| Category | Regex | Meaning | Example |
|:---|:---:|:---|:---|
| Logic | `|` | OR | `cat|dog` |
| Group | `()` | Creates a group | `(cat|dog)` |
| Group | `(?:...)` | Non-capturing group | `(?:cat|dog)` |

---

## 6. Lookahead

| Category | Regex | Meaning | Example |
|:---|:---:|:---|:---|
| Lookahead | `(?=...)` | Followed by | `\w+(?=@)` |
| Lookahead | `(?!...)` | Not followed by | `\w+(?!@)` |

---

## 7. Lookbehind

| Category | Regex | Meaning | Example |
|:---|:---:|:---|:---|
| Lookbehind | `(?<=...)` | Preceded by | `(?<=\$)\d+` |
| Lookbehind | `(?<!...)` | Not preceded by | `(?<!\$)\d+` |

---

## 8. Common Regex Examples

| Requirement | Regex | Meaning |
|:---|:---:|:---|
| Find digits | `\d+` | One or more digits |
| Find 3 digits | `\d{3}` | Exactly 3 digits |
| Find letters | `[A-Za-z]+` | One or more letters |
| Find lowercase letters | `[a-z]+` | One or more lowercase letters |
| Find uppercase letters | `[A-Z]+` | One or more uppercase letters |
| Find special characters | `[^A-Za-z0-9]` | Anything except letters/digits |
| Find non-space characters | `\S+` | One or more non-whitespace characters |
| Find word | `\w+` | One or more word characters |
| Before `@` | `^[^@]+` | Characters from beginning until `@` |
| Username before `@` | `[^@\s]+(?=@)` | Non-space characters immediately before `@` |
| Find email-like text | `[\w.-]+@[\w.-]+\.\w+` | Basic email pattern |
| Find 10-digit number | `\d{10}` | Exactly 10 digits |
| Starts with A | `^A` | Starts with uppercase A |
| Ends with `com` | `com$` | Ends with `com` |

```python
#example 1
#extracting names that starts with capital letter
import re
st = "PhoebeLitchfield is my favorite crickter, MamithaBaiju, KalyaniPriyadarshan are popular in malayalam movies"
s = '[A-Z][a-z]+' #One uppercase letter followed by one or more lowercase letters.
names = re.findall(s,st)
print(names)
```

```python
#example 2
#extracting names from emails
import re
with open('/content/mails.txt','r') as f:
    st = f.read()
print("-"*70)
print("Original File Data:",st)
print("-"*70)
s = r'[^@\s]+(?=@)'
names = re.findall(s,st)
print("extracting names from user emails")
print("-"*70)
print("extracted data")
print(names)
print("-"*70)
```
