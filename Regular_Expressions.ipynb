```python
#finding all occurences of a letter or word
import re
st = "regular expressions are used to find patterns"
s = "a"
lst = re.findall(s,st)
print(lst)
st2 = "lisbon rigsby jane jane jane lisbon Jane rigsby mentalist"
s2 = "jane"
lst2 = re.findall(s2,st2)
print(lst2)
```

```python
st = "lisbon rigsby jane jane jane lisbon Jane rigsby mentalist"
s = "jane"
print("{} found {} times".format(s,len(re.findall(s,st))) if len(re.findall(s,st))>1 else "Count of {} is 0".format(s))
s2 = "robbin"
print("{} found {} times".format(s,len(re.findall(s2,st))) if len(re.findall(s2,st))>1 else "Count of {} is 0".format(s2))
```

```python
#finding occurences of 'b','y'
st = "lisbon rigsby"
s = "[by]"
res = re.finditer(s,st)
for result in res:
  print(result.span())
```

```python
st = "lisbon rigsby"
s = "[by]"
res = re.finditer(s,st)
for result in res:
  print(result.group())
```

```python
#finding indexes of other letter ocuurences except 'aeiou'
import re
st="mamitha"
s="[^aeiou]"
res=re.finditer(s,st)
for result in res:
	print("Start Index:{}  End Index:{}  Value:{}".format(result.start(),result.end(),result.group()))
```

```python
#occurences of capital letters
import re
st = "Kalyani Priyadarshan, Mamitha, Bhagyasri"
s = '[A-Z]'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#occurences of other letters except capitals
import re
st = "KaLyAnI@5/4"
s = '[^A-Z]'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#occurences of lower letters
import re
st = "KaLyAnI"
s = '[a-z]'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#occurences of letters except lower
import re
st = "KaLyAnI@5/4"
s = '[^a-z]'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#occurences of digits
import re
st = "KaLyAnI@5/4"
s = '[0-9]'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#occurences of all characters
st = "KaLyAnI@5/4"
s = '[A-Za-z]'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#occurences of all alphanumerics
st = "KaLyAnI@5/4"
s = '[0-9A-Za-z]'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#occurences of all special characters
st = "KaLyAnI@5/4"
s = '[^0-9A-Za-z]'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#occurences of digits
import re
st = "KaLyAnI@5/4"
s = r'\d'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#occurences of special characters
#\s-used for finding spaces, tabs, new lines
st = "KaLyAnI @5/4  Priy@d@r$h@N"
s = r'\s'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#\w-used for alphanumerics
st = "KaLyAnI @5/4"
s = r'\w'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
#\W - for special chars
st = "KaLyAnI @5/4"
s = r'\W'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
st = "KkalLyyyAaNniIi @5/4/1993"
s = r'\d{4}'
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
st = "KkalLyyyAaNniIi @5/04/1993"
s = r'\d{2,200}' #searches 2 digit number, 201 digits → the regex can match the first 200 digits
res = re.finditer(s,st)
for result in res:
  print(result)
```

```python
st = "KkalLyyyAaNniIi @5/04/1993"
s = r'\d{2,3}' #searches 2 digit number, 6 digits → the regex can match the first 5 digits
res = re.finditer(s,st)
for result in res:
  print(result)
```
